# El valor presente neto

## Introducción

El Valor Presente Neto (VPN), o Valor Actual Neto (VAN), es una herramienta financiera que determina la rentabilidad de un proyecto o inversión. Calcula la diferencia entre el valor actual de los flujos de efectivo futuros ***(ingresos)*** y la inversión inicial, descontados a una tasa de interés específica.

Es favorable por:
  - Indicador de viabilidad
  - Maximización de riqueza
  - Comparación de proyectos



Ventajas del valor presente neto:

1. Utiliza el concepto del valor del dinero en el tiempo
2. Siempre da la desición correcta

### Formula VPN
 $$ VPN = \sum_{t=0}^{n} \frac{C_t}{(1+r)^t} $$
