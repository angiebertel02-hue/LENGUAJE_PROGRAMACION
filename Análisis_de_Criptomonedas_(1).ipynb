```python
# Instalar librerías
!pip install yfinance plotly ta -q

# Importar las librerias
import yfinance as yf
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import ipywidgets as widgets
from IPython.display import display
import plotly.io as pio
```

```python
# Listado de Criptomonedas
criptos = {
    "Bitcoin": "BTC-USD",
    "Ethereum": "ETH-USD",
    "Solana": "SOL-USD",
    "Cardano": "ADA-USD",
    "Litecoin": "LTC-USD",
    "Ripple": "XRP-USD",
    "Polkadot": "DOT-USD",
    "Tether": "USDT-USD",
    "Binance Coin": "BNB-USD",
    "USD Coin": "USDC-USD",
    "Chainlink": "LINK-USD",
    "Dogecoin": "DOGE-USD"}
```

```python
# Widgets Interativos para APP
crypto_dropdown = widgets.Dropdown(
    options=criptos,
    value="BTC-USD",
    description='Criptomoneda:'
)

period_dropdown = widgets.Dropdown(
    options=['7d', '1mo', '3mo', '6mo', '1y'],
    value='1mo',
    description='Período:'
)

interval_dropdown = widgets.Dropdown(
    options=['1h', '1d'],
    value='1h',
    description='Intervalo:'
)

button = widgets.Button(description="Actualizar Datos")
```

## ***Inicio Proceso***

```python
def cargar_datos(b):

    ticker = crypto_dropdown.value
    periodo = period_dropdown.value
    intervalo = interval_dropdown.value

    # Descargar datos
    df = yf.download(
        ticker,
        period=periodo,
        interval=intervalo,
        progress=False,
        auto_adjust=False,
    )

    if df.empty:
        print("No se encontraron datos.")
        return

    # Verificar datos
    #print(df.head())


    # Calcular media móvil
    df['MA20'] = df['Close'].rolling(20).mean()


    # Datos Estadísticos

    precio_actual = float(df['Close'].iloc[-1].iloc[0])
    maximo = float(df['High'].max().iloc[0])
    minimo = float(df['Low'].min().iloc[0])

    rendimiento = float(
    (
        (
            df['Close'].iloc[-1].iloc[0] /
            df['Close'].iloc[0].iloc[0]
        ) - 1
    ) * 100
)

    volatilidad = float(
    df['Close']
    .pct_change()
    .std()
    .values[0]
) * 100

    print("===== ESTADÍSTICAS =====")
    print(f"Precio actual: ${precio_actual:.2f}")
    print(f"Máximo: ${maximo:.2f}")
    print(f"Mínimo: ${minimo:.2f}")
    print(f"Rendimiento: {rendimiento:.2f}%")
    print(f"Volatilidad: {volatilidad:.2f}%")

# =========================================
# EVENTO DEL BOTÓN
# =========================================
button.on_click(cargar_datos)

# =========================================
# MOSTRAR INTERFAZ
# =========================================
display(
    crypto_dropdown,
    period_dropdown,
    interval_dropdown,
    button
)
```
